```python
import os
import sys

os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable

from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("SparkCourse") \
    .master("local[*]") \
    .config("spark.sql.warehouse.dir", "D:/pyspark_udemy_codespace/setup/spark-warehouse") \
    .enableHiveSupport() \
    .getOrCreate()

spark.sparkContext.setLogLevel("ERROR")
print("Spark version:", spark.version)
```

```python
"""
You are given a dataframe below

  df = spark.createDataFrame([("25","07","1982")]).toDF("day", "month", "year")
  Create a date from day, month, and year.
"""
from pyspark.sql.functions import concat_ws, col, to_date # concat_ws stands for concat with seperator

df = spark.createDataFrame([("25","07","1982")]).toDF("day", "month", "year")

date_expr = to_date(concat_ws("-", col("year"), col("month"), col("day")), 'yyyy-MM-dd')
with_date_df = df.withColumn("date", date_expr)
with_date_df.show()
with_date_df.printSchema()
```

```python
"""
You are given a dataframe below

df = (spark.createDataFrame([("Sanjay", "Kalra", 25, "July", 1982, 19408.98)])
          .toDF("fist_name", "last_name", "day", "month", "year", "salary"))

Create the following output

  +-----------------------------------------------+----------+
  |fun_text                                       |salary    |
  +-----------------------------------------------+----------+
  |Mr. Sanjay Kalra was born on 25th July of 1982.|$19,408.98|
  +-----------------------------------------------+----------+
"""
from pyspark.sql.functions import format_number, format_string

df = (spark.createDataFrame([("Sanjay", "Kalra", 25, "July", 1982, 19408.98)])
          .toDF("first_name", "last_name", "day", "month", "year", "salary"))
df.show()

salary_fmt = format_number(col("salary"), 2).alias("salary")
fun_text_fmt = format_string(
    "Mr.%s %s was born on %dth %s of %d.",
    col("first_name"), col("last_name"), col("day"), col("month"), col("year")
).alias("fun_text")

fun_text_df = df.select(fun_text_fmt, salary_fmt)
fun_text_df.show()
```

```python
"""
You are given a dataframe below

  df = spark.createDataFrame([("Benga`li Market", "110001"),("Adu~godi", "560030")]) \
          .toDF("address", "pin")
    Write code to fix the data problem in the address field
"""
from pyspark.sql.functions import translate

df = spark.createDataFrame([("Benga`li Market", "110001"),("Adu~godi", "560030")]).toDF("address", "pin")
df.show()
df.withColumn("address", translate(col("address"), "`~", "")).show() # translate replaces all the charecters in the second param in the first param with the third param
```
