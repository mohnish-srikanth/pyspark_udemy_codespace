```python
spark.sql("CREATE DATABASE IF NOT EXISTS spark_db")
spark.sql("USE spark_db")
print("Database ready")
```

```python
spark.sql("SHOW TABLES IN spark_db").show()
```

```python
import os, shutil

def create_table(name, df):
    path = f"/workspaces/pyspark_udemy_codespace/setup/spark-warehouse/spark_db.db/{name}"
    if os.path.exists(path):
        shutil.rmtree(path)
    spark.sql(f"DROP TABLE IF EXISTS spark_db.{name}")
    df.write.mode("overwrite").saveAsTable(f"spark_db.{name}")
    print(f"Table {name} created OK")
```

```python
create_table("employees", spark.read.json("/workspaces/pyspark_udemy_codespace/data/employees.json"))
create_table("orders", spark.read.json("/workspaces/pyspark_udemy_codespace/data/orders.json"))
create_table("students", spark.read.json("/workspaces/pyspark_udemy_codespace/data/students.json"))
```
