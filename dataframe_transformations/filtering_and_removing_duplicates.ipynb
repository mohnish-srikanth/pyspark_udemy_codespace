```python
"""
Select records that satisfy the follwing criteria
Source is Mumbai and destination is Bangalore

Show the following approaches
    Using a single filter
    Using two filters
    Using a col function expression
    Using a dataframe qualifier expression
"""
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

data_list = [(101, "Mumbai", "Goa", 587),
             (102, "Mumbai", "Bangalore", 985),
             (102, "Mumbai", "Bangalore", 985),
             (103, "Delhi", "Chennai", 2208),
             (104, "Delhi", "Chennai", 2208),
             (105, "Bangalore", "Kolkata", 1868),
             (105, "Bangalore", "Kolkata", 1865)
             ]

data_schema = StructType([
    StructField("id", IntegerType()),
    StructField("source", StringType()),
    StructField("destination", StringType()),
    StructField("distance", IntegerType())
])

df = spark.createDataFrame(data_list, data_schema)
df.show()
```

```python
# using 1 filter
from pyspark.sql.functions import col

df.filter((col("source") == "Mumbai") & (col("destination") == "Bangalore")).show()
```

```python
# using 2 filters

df.filter("source = 'Mumbai'").filter("destination = 'Bangalore'").show()
```

```python
# using dataframe column qualifier

df.filter((df.source == 'Mumbai') & (df.destination == 'Bangalore')).show()
```

```python
# selecting distinct

df.distinct().show()
```

```python
# selecting distinct based on specific columns

df.dropDuplicates(["id", "source", "destination"]).show()
```
