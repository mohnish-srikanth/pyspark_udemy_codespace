```python
# Create DF using Connector
# CSV Connector
file_df = (
    spark.read.format("csv") # csv connector used here
        .option("headers", "true") # connector option
        .option("inferSchema", "true")
        .load(path = "/workspaces/pyspark_udemy_codespace/data/sf-fire-calls.csv")
)
file_df.show(10)
```

```python
# JSON connector
json_df = (
    spark.read.format("json") # json connector used here
        .load(path = "/workspaces/pyspark_udemy_codespace/data/diamonds.json")
)
json_df.show(10)
```

```python
# Create DF usig table
table_df = spark.table("spark_db.diamonds") # table api used to read from table
table_df.show(10)
```

```python
# Create DF using SQL query
sql_df = spark.sql("SELECT * FROM spark_db.diamonds")
sql_df.show(10)
```

```python
# Creating DF from python list
from datetime import datetime, date

data_list_schema = 'id long, name string, joining_date date, salary double, created_at timestamp'

data_list = [(1, "Prashant", date(2018, 1, 1), 924.0, datetime(2022, 1, 1, 9, 0)),
             (2, "Sushant", date(201, 2, 1), 1260.50, datetime(2022, 1, 2, 11, 0)),
             (3, "David", date(2022, 3, 1), 765.0, datetime(2022, 1, 3, 10, 0))]

list_df = spark.createDataFrame(data_list, data_list_schema) # df created using list and list schema
list_df.show()
```

```python
# Create single column DF from range
range_df = spark.range(1000, 1010, 2) # DF created using range spark API
range_df.show()
```
